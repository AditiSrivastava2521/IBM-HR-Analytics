```python
import pandas as pd
```

```python
#Load cleaned dataset
data = pd.read_csv("IBM_HR_Cleaned.csv")
print("Shape:", data.shape)
```

Feature Engineering: Creating Employee Segments

```python
#creating age groups
data['AgeGroup'] = pd.cut(
    data['Age'],
    bins=[0, 24, 34, 44, 54, 100],
    labels=['Under 25', '25-34', '35-44', '45-54', '55+']
)

print(data[['Age', 'AgeGroup']].head(10))
```

```python
#creating Tenure Groups
data['TenureGroup'] = pd.cut(
    data['YearsAtCompany'],
    bins=[-1, 2, 5, 10, 100],
    labels=['0-2', '3-5', '6-10', '10+']
)
print(data[['YearsAtCompany', 'TenureGroup']].head(10))
```

```python
data.to_csv("IBM_HR_Analytical.csv", index=False)
```
