```python
import pandas as pd
from scipy.stats import chi2_contingency
from scipy.stats import ttest_ind
```

```python
data = pd.read_csv("IBM_HR_Analytical.csv")
print("Shape:", data.shape)
```

##Chi-Square Tests


This is to test whether there is a statistically significant association between employee attrition and selected categorical variables.

```python
#function for chi-square test
def chi_square_test(column):
    table = pd.crosstab(data[column], data['Attrition'])
    chi2, p, dof, expected = chi2_contingency(table)

    print("Variable:", column)
    print("Chi-square:", chi2)
    print("p-value:", p)

    if p < 0.05:
        print("Result: Statistically significant association")
    else:
        print("Result: No statistically significant association")
```

```python
chi_square_test('OverTime')
```

Inference: A statistically significant association was found between overtime and employee attrition (p < 0.05).

```python
chi_square_test('JobRole')
```

Inference: Job role has a statistically significant association with employee attrition.

```python
chi_square_test('JobSatisfaction')
```

Inference: Job satisfaction shows a statistically significant association with employee attrition (p < 0.05)

```python
chi_square_test('WorkLifeBalance')
```

Inference: Work-life balance shows a statistically significant association with employee attrition (p < 0.05).

```python
chi_square_test('BusinessTravel')
```

Inference: Business travel has a statistically significant association with employee attrition (p < 0.05).

```python
chi_square_test('AgeGroup')
```

Inference: Age group has a statistically significant association with employee attrition (p < 0.05).

```python
chi_square_test('TenureGroup')
```

Inference: Tenure group has a statistically significant association with employee attrition (p < 0.05).

##T-Test

This determines whether the average values of numerical variables differ significantly between employees who left and those who stayed.

```python
#Monthly Income
left_income = data[data['Attrition'] == 'Yes']['MonthlyIncome']
stayed_income = data[data['Attrition'] == 'No']['MonthlyIncome']

t_stat, p_value = ttest_ind(left_income, stayed_income)

print("T-statistic:", t_stat)
print("p-value:", p_value)
```

Inference: Average monthly income differs significantly between employees who left and those who stayed (p < 0.05).

```python
#Age
left_age = data[data['Attrition'] == 'Yes']['Age']
stayed_age = data[data['Attrition'] == 'No']['Age']

t_stat, p_value = ttest_ind(left_age, stayed_age)

print("T-statistic:", t_stat)
print("p-value:", p_value)
```

Inference: Average age differs significantly between employees who left and those who stayed (p < 0.05).

```python
#Years at Company
left_tenure = data[data['Attrition'] == 'Yes']['YearsAtCompany']
stayed_tenure = data[data['Attrition'] == 'No']['YearsAtCompany']

t_stat, p_value = ttest_ind(left_tenure, stayed_tenure)

print("T-statistic:", t_stat)
print("p-value:", p_value)
```

Inference: Average years at the company differs significantly between employees who left and those who stayed (p < 0.05).

```python
#Years Since Last Promotion
left_promotion = data[data['Attrition'] == 'Yes']['YearsSinceLastPromotion']
stayed_promotion = data[data['Attrition'] == 'No']['YearsSinceLastPromotion']

t_stat, p_value = ttest_ind(left_promotion, stayed_promotion)

print("T-statistic:", t_stat)
print("p-value:", p_value)
```

Inference: YearsSinceLastPromotion does not show a statistically significant difference between employees who left and those who stayed (p > 0.05).

###Final Statistical Summary

#### Key Findings

* All seven categorical factors tested showed a statistically significant association with attrition.
* OverTime and JobRole showed the strongest associations with attrition.
* JobSatisfaction, WorkLifeBalance, and BusinessTravel showed weaker associations.
* Among numerical variables, employees who left were younger, had lower average monthly income, and had fewer years at the company.
* YearsSinceLastPromotion did not show a statistically significant difference between employees who left and those who stayed.

The analysis suggests that work conditions, job role, age, tenure, income, and employee satisfaction are associated with attrition, while time since the last promotion was not statistically significant in this dataset.
